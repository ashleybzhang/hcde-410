*Bring the probe and the observation together into one account. What does this mean for trustworthy, reproducible research — and what will you change about how you use and verify AI in your own work?*

I used AI on one counting tasks and asked one other pesrson to test another counting task that had checkable aswers. For the first one, I asked for an audit of tryout payments, and all four runs found the same two people. In the second, the user asked a a count of Google Form responses, and both models gave the correct number. The key takeaway from these findings was that the answers are reliable, but only in a limited sense. What changed from run to run was the wording and format rather than the result. The three ChatGPT runs in the first test were identical, and Gemini ignored my format. In the second, the Claude runs were the same while ChatGPT’s wording varied.

What this means for trustworthy research: Getting the same answer every time doesn’t prove it’s right. I only knew these answers were correct because I was able to check them against the data. The second test also showed that a correct answer can still be misleading. Claude’s wording made the count look like 8 when the real answer was 11, and the user only caught it by reading more carefully. For research, that means the model’s output is a "claim to check", not a finding. It also means reproducibility depends on writing down the prompt, the model, the version, and the time, since the outputs themselves won’t match exactly. Both tests were small and easy, but I'm skeptical to trust AI in how it would hold up on messier data.
